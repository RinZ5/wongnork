```python
import os
import shutil
import kagglehub
```

```python
project_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
destination_dir = os.path.join(project_root, "data", "raw")

if os.path.exists(destination_dir):
    print("Path to dataset files:", destination_dir)
else:
    source = kagglehub.dataset_download("irkaal/foodcom-recipes-and-reviews")
    
    os.makedirs(os.path.dirname(destination_dir), exist_ok=True)
    
    shutil.move(source, destination_dir)
    
    print("Path to dataset files:", destination_dir)
```
